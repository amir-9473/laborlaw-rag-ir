# 04 — Query Processing

**Purpose:** Normalize a Persian question and optionally create LLM-based retrieval variants.

The pipeline owns query preparation so the notebook does not duplicate normalization or transformation logic.


## 1. Setup


```python
from laborlaw_rag.pipeline import RAGPipeline

QUESTION = "شرایط پرداخت حق سنوات به کارگر چیست؟"

# Enable this only when extra LLM latency is acceptable.
USE_QUERY_TRANSFORMATION = False
pipeline = RAGPipeline.from_settings()
```

## 2. Prepare Retrieval Queries


```python
normalized_query, retrieval_queries, query_warnings, transformation_applied = (
    pipeline.prepare_queries(
        QUESTION,
        use_query_transformation=USE_QUERY_TRANSFORMATION,
    )
)
```
