# 06 — End-to-End RAG Pipeline

**Purpose:** Accept a Persian question and print the grounded answer with numbered citations and source references.

The notebook delegates the complete workflow to `laborlaw_rag.pipeline.RAGPipeline`.


## 1. Setup


```python
from laborlaw_rag.pipeline import RAGPipeline

# Enable this only when extra LLM latency is acceptable.
USE_QUERY_TRANSFORMATION = False
```

## 2. Run the Pipeline


```python
question = input("سؤال خود را درباره قانون کار وارد کنید: ")
question = question.strip()

pipeline = RAGPipeline.from_settings()
result = pipeline.ask(
    question,
    use_query_transformation=USE_QUERY_TRANSFORMATION,
)
print(result.final_output)
```
